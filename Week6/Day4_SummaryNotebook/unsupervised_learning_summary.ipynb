# Week 6: Unsupervised Learning Summary
This notebook reviews clustering, scaling, PCA, and anomaly detection.

## 1. KMeans Clustering
KMeans groups similar records without target labels. The elbow method helps choose K by plotting inertia.

```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
data = load_iris().data
scaled = StandardScaler().fit_transform(data)
labels = KMeans(n_clusters=3, random_state=42, n_init=10).fit_predict(scaled)
print('Cluster sizes:', __import__('numpy').bincount(labels))
```

## 2. PCA
PCA reduces many features into fewer principal components while retaining most of the variance.

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
reduced = pca.fit_transform(scaled)
print('New shape:', reduced.shape)
print('Variance retained:', round(pca.explained_variance_ratio_.sum() * 100, 2), '%')
```

## 3. Anomaly Detection
Z-scores measure distance from the mean. IQR flags values below Q1 - 1.5×IQR or above Q3 + 1.5×IQR.

```python
import numpy as np
values = np.array([42, 45, 47, 43, 46, 44, 120])
q1, q3 = np.percentile(values, [25, 75])
lower, upper = q1 - 1.5 * (q3 - q1), q3 + 1.5 * (q3 - q1)
print('IQR anomalies:', values[(values < lower) | (values > upper)])
```

## Key Learnings
- Scale features before distance-based algorithms.
- Use the elbow method and cluster interpretation together.
- PCA makes high-dimensional data easier to visualize.
- Investigate outliers before removing or capping them.
